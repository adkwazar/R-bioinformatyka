Define a function that takes two vectors and computes the cosine of the angle between them (cosine similarity).

The cosine similarity between two vectors $x$ and $y$ is defined as:

$$
\cos(\theta) = \frac{x \cdot y}{\|x\|\|y\|}
$$

where:

- $x \cdot y$ — dot product of vectors  
- $\|x\|$ — Euclidean norm of vector $x$

---

1. Implement a function that computes cosine similarity between two vectors.

2. Check whether the vectors

$$
x_1 = [3, 1, -2, 4], \quad x_2 = [-1, 3, 4, 2]
$$

are orthogonal.

3. Consider the following feature vectors:

$$
x_1 = [3,1,-2,4]
$$

$$
x_2 = [-1,3,4,2]
$$

$$
x_3 = [2,0,-1,3]
$$

$$
x_4 = [-2,4,3,1]
$$

$$
x_5 = [1,1,-1,2]
$$

Each coordinate represents a feature describing an object.

- Compute cosine similarity for **all pairs of vectors**.

- Determine **which pair of objects is the most similar**.

- Determine **which pair of objects is the least similar**.

- Identify pairs of vectors that are **orthogonal or nearly orthogonal**.

```python
def cos_sim(x, y):
    return np.dot(x,y)/(np.linalg.norm(x)*np.linalg.norm(y))

#2
x1 = np.array([3,1,-2,4])
x2 = np.array([-1,3,4,2])

print(cos_sim(x1,x2))

#3
x = [np.array([3,1,-2,4]), np.array([-1,3,4,2]), np.array([2,0,-1,3]), np.array([-2,4,3,1]), np.array([1,1,-1,2])]
for i in range(5):
    for j in range(5):
        print(f"sim between x{i} and x{j}: {cos_sim(x[i],x[j])}")
```

### Exercise10

- Define a PDF for 2D normal distribution given by $f(X) = \frac{1}{2\pi\sqrt{\det(\Sigma)}}\cdot \exp(-\frac{1}{2}(X-\mu)^T\Sigma^{-1}(X-\mu))$, where $X = [X_1, X_2]$, $\mu = [\mu_1,\mu_2]$ and $\Sigma$ is a covariance matrix. Both $\mu$ and $\Sigma$ are parameters.

<br>

- Assuming $\mu=[1,2]$ and $$\Sigma = \begin{bmatrix} 1 & 1.1 \\ 1.1 & 4  \end{bmatrix}$$ Calculate $f(0,1)$.

```python
mu = np.array([1,2])
X = np.array([0,1])
Sigma = np.array([[1,1.1],[1.1,4]])

def normal2d(mu, X, Sigma):
    diff = X - mu
    d = np.linalg.det(Sigma)
    inv = np.linalg.inv(Sigma)
    return 1/(2*np.pi*np.sqrt(d)) * np.exp(-0.5 * np.dot(np.dot(diff.T,  inv),  diff))

print(normal2d(mu, X, Sigma))
```
